#Question 1: When it comes to asking for loans, is there a difference between the gender and the sector?


```python
gender_loans = loans[
    loans['borrower_genders'].isin(['female', 'male'])
]
```

```python
gender_loans[gender_loans['borrower_genders'] == 'female'].groupby('sector').size().sort_values(ascending=False)
```

```python
gender_loans[gender_loans['borrower_genders'] == 'male'].groupby('sector').size().sort_values(ascending=False)
```

```python
female_sector = gender_loans[
    gender_loans['borrower_genders'] == 'female'
].groupby('sector').size()

female_percentage = female_sector / female_sector.sum() * 100

female_percentage.sort_values(ascending=False).round(2)
```

```python
male_sector = gender_loans[
    gender_loans['borrower_genders'] == 'male'
].groupby('sector').size()

male_percentage = male_sector / male_sector.sum() * 100

male_percentage.sort_values(ascending=False).round(2)
```

**Answer:** There is a difference between the genders in relation to the sectors.
Agriculture represents 35.73% of male loan records, compared with 23.35% of female records. Food represents 23.16% of female records, compared with 10.70% of male records. Retail: represents 22.63% of female records, compared with 10.60% of male records.

#Question 4: How do loan amounts differ across the currencies and how are they distributed small and large categories?

```python
loans['currency'].value_counts().head(20)
```

```python
currency_analysis = loans.groupby('currency')['loan_amount'].agg(
    ['count', 'mean']
)
```

```python
currency_analysis = currency_analysis[
    currency_analysis['count'] >= 1000
]
```

```python
currency_analysis.sort_values(
    by='mean',
    ascending=False
)
```

```python
loans['loan_category'] = loans['loan_amount'].apply(
    lambda x: 'Small' if x < 500 else 'Large'
)
```

```python
currency_categories = loans.groupby(
    ['currency', 'loan_category']
).size().reset_index(name='number_of_loans')

currency_categories
```

**Answer:** The results show that average loan amounts differ across the currencies recorded in the dataset. Some currency categories have higher average loan amounts than others.The results show that the number of small and large loans varies across currencies. In most currency categories, large loans are more numerous than small loans. This suggests that the size distribution of loan requests is not the same across all countries and currency categories.

#Question 5: Is there a relationship between how the loans are used and the gender of the borrowers?

```python
female_purposes = loans[
    loans['borrower_genders'] == 'female'
]['use'].value_counts().head(15)

female_purposes
```

```python
male_purposes = loans[
    loans['borrower_genders'] == 'male'
]['use'].value_counts().head(15)

male_purposes
```

```python
female_purposes_percentage = loans[
    loans['borrower_genders'] == 'female'
]['use'].value_counts(normalize=True).head(15) * 100

female_purposes_percentage.round(2)
```

```python
male_purposes_percentage = loans[
    loans['borrower_genders'] == 'male'
]['use'].value_counts(normalize=True).head(15) * 100

male_purposes_percentage.round(2)
```

```python
farm_purposes = loans[
    loans['use'].str.contains('farm', case=False, na=False)
]

farm_purposes['use'].value_counts().head(10)
```

**Answer:** The results suggest differences in the most frequently recorded loan purposes between female and male borrowers. Female borrower records frequently include purposes such as sanitation, food production and livestock, while male borrower records often include agricultural inputs and household needs.
To explore agricultural purposes further, I used `.str.contains()` to identify loan descriptions containing the word "farm". This allowed me to examine examples of farming-related loan purposes in the dataset.



#Question 6: Based on the genders, which interval are they most likely to repay the amount?

```python
female_repayment = loans[
    loans['borrower_genders'] == 'female'
]['repayment_interval'].value_counts()

female_repayment
```

```python
male_repayment = loans[
    loans['borrower_genders'] == 'male'
]['repayment_interval'].value_counts()

male_repayment
```

```python
female_repayment_percentage = female_repayment / female_repayment.sum() * 100

female_repayment_percentage.round(2)
```

```python
male_repayment_percentage = male_repayment / male_repayment.sum() * 100

male_repayment_percentage.round(2)
```

**Answer:**
- Female loan records are most frequently associated with irregular repayments (49.25%).
- Male loan records are most frequently associated with monthly repayments (65.04%).
- Bullet repayment represents a larger proportion of male records (21.58%) than female records (5.19%).
- Weekly repayment is uncommon for both groups.
